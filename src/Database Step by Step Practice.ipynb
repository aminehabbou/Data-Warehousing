### Display the number of customers per country

```python
conn=sqlite3.connect(database_path_filename)

cur=conn.cursor()

cur.execute("SELECT Country, COUNT(*) AS Numbers FROM Customer GROUP BY Country ORDER BY Numbers DESC")

result=cur.fetchall()

for i in result:
    print(i)

conn.close()
```

### Display the number of tracks per artists

```python
conn=sqlite3.connect(database_path_filename)

cur=conn.cursor()

cur.execute("""SELECT Artist.Name, COUNT (Track.TrackId) AS track_count
            FROM Artist
            LEFT JOIN Album ON Artist.ArtistId=Album.ArtistId
            LEFT JOIN Track ON Album.AlbumId=Track.AlbumId
            GROUP BY Artist.Name
            ORDER BY track_count DESC
            """)

result=cur.fetchall()

for i in result:
    print(i)

conn.close()
```

### Display the number of customers per country

```python
conn=sqlite3.connect(database_path_filename)
cur=conn.cursor()

cur.execute("""SELECT Customer.Country, Count (Customer.CustomerId) AS customer_count_per_country
            FROM Customer
            GROUP BY Customer.Country
            ORDER BY customer_count_per_country DESC
             """)

result=cur.fetchall()

for i in result:
    print(i)
    
conn.close()
```

### Find Products Never Purchased

```python
conn=sqlite3.connect(database_path_filename)

cur=conn.cursor()

cur.execute("""SELECT Track.Name, Track.UnitPrice
            FROM Track
            LEFT JOIN InvoiceLine ON Track.TrackId=InvoiceLine.TrackId
            WHERE InvoiceLineId IS NULL
            """)

result=cur.fetchall()

for i in result:
    print(i)

conn.close()
```

### Find the customer who has spent the most money

```python
conn=sqlite3.connect(database_path_filename)
cur=conn.cursor()

cur.execute("""SELECT Customer.LastName, SUM(Invoice.Total) AS total_spent
            FROM Customer
            LEFT JOIN Invoice ON Customer.CustomerId=Invoice.CustomerId
            GROUP BY Customer.LastName
            ORDER BY total_spent DESC
            LIMIT 1
            """)

result=cur.fetchall()

for i in result:
    print(i)
    
conn.close()
```

### Calculate total spent per customer

```python
conn=sqlite3.connect(database_path_filename)
cur=conn.cursor()

cur.execute("""SELECT Customer.LastName, SUM(Invoice.Total) AS total_spent
            FROM Customer
            LEFT JOIN Invoice ON Customer.CustomerId=Invoice.CustomerId
            GROUP BY Customer.LastName
            ORDER BY total_spent DESC
            """)

result=cur.fetchall()

for i in result:
    print(i)
    
conn.close()
```

### Find the customer who has made the least purchases (invoices)

```python
conn=sqlite3.connect(database_path_filename)

cur=conn.cursor()

cur.execute("""SELECT Customer.LastName, COUNT(Invoice.InvoiceId) as invoices_count
            FROM Customer
            LEFT JOIN Invoice ON Customer.CustomerId=Invoice.CustomerId
            GROUP BY Customer.LastName
            ORDER BY invoices_count ASC
            LIMIT 1
            """)

result=cur.fetchall()

for i in result:
    print(i)

conn.close()
```

### List all customers and their total spending, sorted by spending

```python
conn=sqlite3.connect(database_path_filename)
cur=conn.cursor()

cur.execute("""SELECT Customer.LastName, SUM(Invoice.Total) AS total_spent
            FROM Customer
            LEFT JOIN Invoice ON Customer.CustomerId=Invoice.CustomerId
            GROUP BY Customer.LastName
            ORDER BY total_spent DESC
            """)

result=cur.fetchall()

for i in result:
    print(i)
    
conn.close()
```

### Find the album with the most tracks

```python
conn=sqlite3.connect(database_path_filename)
cur=conn.cursor()

cur.execute("""SELECT Album.Title, COUNT(Track.TrackId) as track_count
            FROM Album
            LEFT JOIN Track ON Album.AlbumId=Track.AlbumId
            GROUP BY Album.Title
            ORDER BY track_count DESC
            LIMIT 1
            """)

result=cur.fetchall()

for i in result:
    print(i)
    
conn.close()
```

### List all artists who have released more than 10 albums

```python
conn=sqlite3.connect(database_path_filename)
cur=conn.cursor()

cur.execute("""SELECT Artist.Name, COUNT(Album.AlbumId) AS album_count
            FROM Artist
            LEFT JOIN Album ON Artist.ArtistId=Album.ArtistId
            GROUP BY Artist.Name
            HAVING album_count > 10
            ORDER BY album_count DESC
            """)

result=cur.fetchall()

for i in result:
    print(i)
    
conn.close()
```

### Find the average price of a track in each genre

```python
conn=sqlite3.connect(database_path_filename)

cur=conn.cursor()

cur.execute("""SELECT Genre.Name, AVG(Track.UnitPrice) as Track_avgprice
            FROM Genre
            LEFT JOIN Track ON Genre.GenreId=Track.GenreId
            GROUP BY Genre.GenreId
            """)

result=cur.fetchall()

for i in result:
    print(i)

conn.close()
```
